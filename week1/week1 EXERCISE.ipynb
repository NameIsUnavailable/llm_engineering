```python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
# set up environment
# load_dotenv(override=True)
# api_key = os.getenv('OPENAI_API_KEY')

# #Check the key
# if not api_key:
#    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
# elif not api_key.startswith("sk-proj-"):
#     print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
# elif api_key.strip() != api_key:
#     print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
# else:
#     print("API key found and looks good so far!")

#Use Python CLient Library 
openai = OpenAI()

tutor_system_prompt = """
You will be a tutor who is an expert in techincal terms for subjects related to IT, computer science,
data science, cybersecurity, and other related domains. Explain and clear up misunderstandings and add
clarification through examples, analogies, and reframing answers if necessary. Breakdown each point and
mention vocabulary words that are necessary to understand more about the question. Answers do not have to 
be very long unless otherwise asked. Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.  
"""

tutor_user_prompt = """ 
I need help understanding unfamiliar technical terms. I am still learning and trying get a better understanding since
there are so much knowledge and terms that one has to know already and it is a bit overwhelming.

"""
```

```python
# here is the question; type over this to ask something new

question = """
How does github work? I am still confused about push, pull, and how exactly I push code into my repo from 
my computer. Do I need to have constant connection to my repo or something?  
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

def GPTtutorResponse():
    stream = openai.chat.completions.create(
        model= MODEL_GPT,
        messages=[
            {"role": "system", "content": tutor_system_prompt},
            {"role": "user", "content": tutor_user_prompt + question}
          ],
        stream=True
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
GPTtutorResponse()
```

```python
# Get Llama 3.2 to answer

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
response = ollama.chat.completions.create(
    model = MODEL_LLAMA, 
    messages =[{"role": "system", "content": tutor_user_prompt},
               {"role": "user", "content": tutor_user_prompt + question}] 
                )
tutorSummary = response.choices[0].message.content
print(display(Markdown(tutorSummary)))
```
